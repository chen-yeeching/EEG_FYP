```python
# -------------------------------
# 0. IMPORTS
# -------------------------------
import pandas as pd
import numpy as np
import glob, os, warnings
import matplotlib.pyplot as plt
import tensorflow as tf

from collections import Counter

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
from sklearn.utils.class_weight import compute_class_weight
from sklearn.model_selection import StratifiedShuffleSplit, train_test_split

from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.layers import Input, Conv2D, DepthwiseConv2D, SeparableConv2D, BatchNormalization, Activation, AveragePooling2D, GlobalAveragePooling2D, Dropout, Flatten, Dense, Reshape, Permute, Multiply
from tensorflow.keras.regularizers import l2

warnings.simplefilter(action='ignore', category=pd.errors.PerformanceWarning)
warnings.filterwarnings("ignore")

# -------------------------------
# 1. CONFIG / CONSTANTS
# -------------------------------
FOLDER_PATH = '/content/drive/MyDrive/EEG_Data/Final Data/'
EMOTION_LABELS = {'Happy': 0, 'Neutral': 1, 'Sad': 2, 'Anxiety': 3}

TARGET_SENSORS = [
    'Cz', 'Fz', 'Fp1', 'F7', 'F3', 'FC1', 'C3', 'FC5',
    'FT9', 'T7', 'CP5', 'CP1', 'P3', 'P7', 'PO9', 'O1',
    'Pz', 'Oz', 'O2', 'PO10', 'P8', 'P4', 'CP2', 'CP6',
    'T8', 'FT10', 'FC6', 'C4', 'FC2', 'F4', 'F8', 'Fp2'
]

WINDOW_SIZE = 128
STEP_SIZE = 64    # 50% overlap
MAX_WINDOWS_PER_FILE = 30

# -------------------------------
# 2. HELPER FUNCTIONS
# -------------------------------
def load_csv_with_auto_header(filepath):
    """Load CSV, try header row 1, fallback to row 0."""
    df = pd.read_csv(filepath, header=1)
    if not any(col.startswith("EEG.") for col in df.columns.astype(str)):
        df = pd.read_csv(filepath, header=0)
    return df

def get_eeg_columns(df):
    """Return EEG column names and corresponding channel names."""
    eeg_cols = []
    ch_names = []

    df_cols_upper = [c.upper() for c in df.columns.astype(str)]

    for ch in TARGET_SENSORS:
        col_name = None
        # Look for a column in df that matches 'EEG.<ch>'
        for c_orig, c_up in zip(df.columns.astype(str), df_cols_upper):
            if c_up == f"EEG.{ch.upper()}":
                col_name = c_orig
                break
        if col_name is not None:
            eeg_cols.append(col_name)
            ch_names.append(ch.upper())

    return eeg_cols, ch_names

def load_eeg_segments(
    data_file,
    label_name,
    file_id,
    window_size=WINDOW_SIZE,
    step_size=STEP_SIZE,
):
    try:
        df = load_csv_with_auto_header(data_file)
    except Exception as e:
        print(f"Failed to load {data_file}: {e}")
        return [], []

    eeg_cols, ch_names = get_eeg_columns(df)

    eeg_data = df[eeg_cols].values  # (samples, channels)

    # Normalize per recording
    eeg_data = (eeg_data - np.mean(eeg_data, axis=0)) / (np.std(eeg_data, axis=0) + 1e-6)

    # Sliding window segmentation
    X_segments, y_segments, file_ids = [], [], []
    n_samples = eeg_data.shape[0]
    for start in range(0, n_samples - window_size + 1, step_size):
      if len(X_segments) >= MAX_WINDOWS_PER_FILE:
          break
      segment = eeg_data[start:start + window_size, :]
      X_segments.append(segment)
      y_segments.append(EMOTION_LABELS[label_name])
      file_ids.append(file_id)

    return X_segments, y_segments, file_ids

def split_by_file(X, y, file_ids):
    unique_files = np.unique(file_ids)

    # Get label per file
    file_labels = np.array([y[np.where(file_ids == f)[0][0]] for f in unique_files])

    # Split files into train_val (75%) and test (25%)
    sss_test = StratifiedShuffleSplit(n_splits=1, test_size=0.25, random_state=42)
    train_val_idx, test_idx = next(sss_test.split(unique_files, file_labels))
    train_val_files = unique_files[train_val_idx]
    test_files = unique_files[test_idx]

    # Labels for train_val files
    train_val_labels = file_labels[train_val_idx]

    # Split train_val into train (5 per emotion) and validation (1 per emotion)
    sss_val = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
    train_idx, val_idx = next(sss_val.split(train_val_files, train_val_labels))
    train_files = train_val_files[train_idx]
    val_files = train_val_files[val_idx]

    # Create masks for X/y selection
    train_mask = np.isin(file_ids, train_files)
    val_mask = np.isin(file_ids, val_files)
    test_mask = np.isin(file_ids, test_files)

    return (
        X[train_mask], y[train_mask],
        X[val_mask], y[val_mask],
        X[test_mask], y[test_mask]
    )

def augment_segment(segment, noise_std=0.01, shift_max=5, scale_range=(0.95, 1.05)):
    """Apply noise, shift, and amplitude scaling to a segment."""
    seg = segment.copy()

    # Gaussian noise
    seg += np.random.normal(0, noise_std, seg.shape)

    # Temporal shift
    shift = np.random.randint(-shift_max, shift_max)
    if shift > 0:
        seg = np.pad(seg, ((shift, 0), (0, 0)), mode='constant')[:-shift]
    elif shift < 0:
        seg = np.pad(seg, ((0, -shift), (0, 0)), mode='constant')[-shift:]

    # Amplitude scaling
    scale = np.random.uniform(*scale_range)
    seg *= scale

    return seg

def augment_to_balance(X, y, max_ratio=2.0, sad_ratio=2.3, max_aug_per_segment=3):
    counts = Counter(y)
    max_count = int(max(counts.values()) * max_ratio)

    X_new, y_new = list(X), list(y)
    augmentation_tracker = {i: 0 for i in range(len(X))}  # track augmentations per segment

    for cls in counts:
        idxs = np.where(y == cls)[0]

        # Determine target count
        if cls == 2:  # Sad
            target_count = int(max_count * sad_ratio / max_ratio)
        else:
            target_count = max_count

        while counts[cls] < target_count:
            i = np.random.choice(idxs)
            # Skip if segment has been augmented too many times
            if augmentation_tracker[i] >= max_aug_per_segment:
                continue
            # Randomize augmentation parameters slightly
            if cls == 2:
                noise = np.random.uniform(0.01, 0.03)
                shift = np.random.randint(3, 7)
                scale = (0.90, 1.10)
            else:
                noise = np.random.uniform(0.005, 0.02)
                shift = np.random.randint(2, 5)
                scale = (0.95, 1.05)

            X_new.append(augment_segment(X[i], noise_std=noise, shift_max=shift))
            y_new.append(cls)
            counts[cls] += 1
            augmentation_tracker[i] += 1

    return np.array(X_new), np.array(y_new)

def channel_attention(x, ratio=8):
    channels = x.shape[-1]

    se = GlobalAveragePooling2D()(x)
    se = Dense(channels // ratio, activation='relu')(se)
    se = Dense(channels, activation='sigmoid')(se)
    se = Reshape((1, 1, channels))(se)

    return Multiply()([x, se])

def build_eegnet(input_shape=(128, 32), n_classes=4):
    inputs = Input(shape=input_shape)
    x = Reshape((input_shape[0], input_shape[1], 1))(inputs)

    # Temporal convolution
    x = Conv2D(16, (32, 1), padding='same', kernel_regularizer=l2(1e-4), use_bias=False)(x)
    x = BatchNormalization()(x)
    x = Activation('elu')(x)

    # Depthwise spatial convolution
    x = DepthwiseConv2D((1, input_shape[1]), depth_multiplier=2, depthwise_regularizer=l2(1e-4), use_bias=False)(x)
    x = BatchNormalization()(x)
    x = Activation('elu')(x)
    x = AveragePooling2D((4, 1))(x)
    x = Dropout(0.5)(x)

    # Separable convolution
    x = SeparableConv2D(32, (16, 1), padding='same', depthwise_regularizer=l2(1e-4), pointwise_regularizer=l2(1e-4), use_bias=False)(x)
    x = BatchNormalization()(x)
    x = Activation('elu')(x)

    x = channel_attention(x)

    x = AveragePooling2D((8, 1))(x)
    x = Dropout(0.5)(x)

    # Classifier
    x = Flatten()(x)
    outputs = Dense(n_classes, activation='softmax')(x)

    return Model(inputs, outputs)

def focal_loss(alpha=None, gamma=2.0):
    def loss(y_true, y_pred):
        y_true = tf.cast(y_true, tf.int32)
        y_true_onehot = tf.one_hot(y_true, depth=tf.shape(y_pred)[-1])
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1.0 - 1e-7)

        cross_entropy = -y_true_onehot * tf.math.log(y_pred)
        weight = tf.pow(1 - y_pred, gamma)

        if alpha is not None:
            alpha_tensor = tf.constant(alpha, dtype=tf.float32)
            weight *= alpha_tensor

        return tf.reduce_sum(weight * cross_entropy, axis=-1)
    return loss

# -------------------------------
# 3. LOAD ALL DATA
# -------------------------------
X_all, y_all, file_id_all = [], [], []

data_files = glob.glob(os.path.join(FOLDER_PATH, "*md.bp.csv"))

for data_file in data_files:
    filename = os.path.basename(data_file)
    label_name = None
    for label in EMOTION_LABELS.keys():
        if label in filename: label_name = label
    if label_name is None: continue

    # Load EEG segments
    X_seg, y_seg, file_ids = load_eeg_segments(
        data_file,
        label_name,
        file_id=filename,
    )

    if len(X_seg) == 0:
        continue  # skip if no segments

    X_all.extend(X_seg)
    y_all.extend(y_seg)
    file_id_all.extend(file_ids)

# Convert to numpy arrays
X_all = np.array(X_all, dtype=np.float32)
y_all = np.array(y_all)
file_id_all = np.array(file_id_all)
print(f"\nTotal segments loaded: {len(X_all)}")

# -------------------------------
# 4. SPLIT TRAIN/VAL/TEST
# -------------------------------
X_train, y_train, X_val, y_val, X_test, y_test = split_by_file(X_all, y_all, file_id_all)

# Augment training data
X_train, y_train = augment_to_balance(X_train, y_train)

# -------------------------------
# 5. BUILD CNN MODEL
# -------------------------------
input_shape = X_train.shape[1:]  # (time_steps, channels)
n_classes = len(EMOTION_LABELS)
alpha = [1.0, 1.0, 1.3, 1.0]
model = build_eegnet(input_shape, n_classes)
model.compile(optimizer='adam', loss=focal_loss(alpha=alpha, gamma=2.0), metrics=['accuracy'])
model.summary()

# -------------------------------
# 6. TRAIN
# -------------------------------
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True,
    verbose=1
)

lr_scheduler = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-5,
    verbose=1
)

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[early_stop, lr_scheduler],
    verbose=2
)

# -------------------------------
# 7. EVALUATE
# -------------------------------
y_pred = np.argmax(model.predict(X_test), axis=1)

# Per-class metrics
print("\n--- Classification Report ---")
print(classification_report(y_test, y_pred, target_names=EMOTION_LABELS.keys(), zero_division=0))

macro_f1 = f1_score(y_test, y_pred, average='macro')
print(f"Macro F1-score: {macro_f1:.2f}")

overall_acc = np.mean(y_test == y_pred)
print(f"Overall Model Accuracy: {overall_acc:.2f}")

# Confusion matrix (normalized)
cm = confusion_matrix(y_test, y_pred, normalize='true')
display_labels = list(EMOTION_LABELS.keys())
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=display_labels)
disp.plot(cmap=plt.cm.Blues)
plt.title("CNN Confusion Matrix")
plt.show()

# -------------------------------
# 8. PLOT LOSS/ACCURACY
# -------------------------------
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.legend()
plt.title('Loss')

plt.subplot(1,2,2)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.legend()
plt.title('Accuracy')
plt.show()
```
